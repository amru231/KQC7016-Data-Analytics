
# Exploratory Data Visualization using Seaborn

This notebook demonstrates how to visualize data effectively using Seaborn.  
We'll use a synthetic dataset representing employee productivity metrics.


```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("employee_productivity.csv")
df.head()
```

## 1. Univariate Visualizations

```python

# Histogram of Salary
sns.histplot(df['Salary'], kde=True, color='skyblue')
plt.title('Distribution of Salary')
plt.show()

# Boxplot of Salary by Department
sns.boxplot(x='Department', y='Salary', data=df)
plt.title('Salary Distribution across Departments')
plt.show()

# Countplot of Departments
sns.countplot(x='Department', hue='Department', data=df, palette='Set2', legend=False)
plt.title('Employee Count per Department')
plt.show()
```

## 2. Bivariate Visualizations

```python

# Scatterplot: Experience vs Salary
sns.scatterplot(x='Experience_Years', y='Salary', hue='Department', data=df)
plt.title('Experience vs Salary by Department')
plt.show()

# Lineplot: Age vs Salary
sns.lineplot(x='Age', y='Salary', data=df, errorbar=None)
plt.title('Average Salary vs Age')
plt.show()

# Barplot: Department vs Average Performance
sns.barplot(x='Department', y='Performance_Score', hue='Department', data=df, \
            errorbar=None, palette='muted', legend=False )
plt.title('Average Performance Score by Department')
plt.show()
```

## 3. Multivariate Visualizations

```python
# Pairplot among numeric variables
sns.pairplot(df[['Age','Experience_Years','Salary','Performance_Score']], diag_kind='kde')
plt.show()

# Heatmap of correlations
corr = df[['Age','Experience_Years','Salary','Performance_Score','Satisfaction_Level','Work_Hours']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()

# Violin plot: Work hours vs Satisfaction Level
sns.violinplot(x='Overtime', y='Satisfaction_Level', data=df, hue='Overtime', palette='pastel', legend=False)
plt.title('Satisfaction Level by Overtime')
plt.show()
```


## Summary
In this notebook, we visualized different aspects of the employee dataset using Seaborn:
- **Histograms and boxplots** to view distributions  
- **Scatterplots and lineplots** to explore relationships  
- **Pairplots and heatmaps** for multivariate patterns  
Visualization helps uncover insights before moving into modeling or hypothesis testing.
